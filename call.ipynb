# Blog Post Title Generation Example Call

```python
import requests

def generate_blog_titles(content: str, api_key: str, style: str = 'descriptive') -> dict:
    url = "http://localhost:8000/blog_title/posts/generate_titles/"
    
    data = {
        "content": content,
        "openai_key": api_key,
        "style": style,
        "max_titles": 3
    }
    
    try:
        response = requests.post(url, json=data)
        response.raise_for_status()
        return response.json()
    except requests.exceptions.RequestException as e:
        print(f"Error: {str(e)}")
        return None
```
